# 01 External DB setup

Seeds CultPass (`data/external/cultpass.db`).

```python
from setup_external_db import setup_cultpass_db
from sqlalchemy import create_engine
from utils import get_session
from data.models import cultpass

db_path = setup_cultpass_db()
engine = create_engine(f"sqlite:///{db_path}", echo=False)
```

```python
with get_session(engine) as session:
    for user in session.query(cultpass.User).all():
        print(user, user.subscription)
```

```python
with get_session(engine) as session:
    for experience in session.query(cultpass.Experience).all():
        print(experience)
```
