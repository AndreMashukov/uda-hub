# 02 Core DB setup

Seeds UDA-Hub (`data/core/udahub.db`) including at least 14 CultPass articles.

```python
from setup_core_db import load_articles, setup_udahub_db
from sqlalchemy import create_engine
from utils import get_session
from data.models import udahub

assert len(load_articles()) >= 14
db_path = setup_udahub_db()
engine = create_engine(f"sqlite:///{db_path}", echo=False)
```

```python
with get_session(engine) as session:
    account = session.query(udahub.Account).filter_by(account_id="cultpass").first()
    print(account)
    print("articles", len(account.knowledge_articles))
    for article in account.knowledge_articles:
        print(article)
```
